```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv(r'C:\Users\Dell Precison\Documents\Python\Data Analysis\healthcare_dataset.csv')

# 2. Inspect missing values and structural data types
print(df.info())

# 3. Data Cleaning
# Standardize text columns (strip whitespace, enforce proper capitalization)
df['Medical Condition'] = df['Medical Condition'].str.strip().str.title()
df['Insurance Provider'] = df['Insurance Provider'].str.strip().str.title()

# Handle negative or zero billing amounts if anomalies exist
df['Billing Amount'] = df['Billing Amount'].apply(lambda x: x if x > 0 else np.nan)
df['Billing Amount'] = df['Billing Amount'].fillna(df['Billing Amount'].median())

# Round financial figures to two decimal places
df['Billing Amount'] = df['Billing Amount'].round(2)

# 4. Feature Engineering (Crucial for BI Visuals)
# Create age groups to make dashboard filtering easier
age_bins = [0, 18, 35, 50, 65, 100]
age_labels = ['0-17', '18-34', '35-49', '50-64', '65+']
df['Age Group'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels)

# 5. Export clean data for Tableau and Power BI
df.to_csv(r'C:\Users\Dell Precison\Documents\Python\Data Analysis\cleaned_healthcare_data2.csv', index=False)
print("Data successfully cleaned and exported!")
```
